# Leitura do Lakehouse com Pandas

Leitura direta do Parquet particionado com `pandas` + `pyarrow` — sem Spark. O pyarrow entende o particionamento Hive (`ingest_date=...`) e devolve a coluna de partição automaticamente.

```python
import pandas as pd

LAKEHOUSE = "/home/jovyan/lakehouse"
```

## veiculos

```python
SOURCE = 'veiculos'
df_raw = pd.read_parquet(f"{LAKEHOUSE}/raw/{SOURCE}", engine="pyarrow")
print(f"raw: {df_raw.shape[0]} linhas x {df_raw.shape[1]} colunas")
```

```python
df_raw['km_atual'] = df_raw['km_atual'].astype(int)
```

```python
df_raw.info()
```
